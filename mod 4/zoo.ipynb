16.4 Use the sqlite3 module to create a SQLite database called books.db and a table called books with these fields: title (text), author (text), and year (integer).


```python
import sqlite3
```

```python
conn = sqlite3.connect("books.db")
```

```python
conn.execute('''CREATE TABLE books(title VARCHAR(100), author VARCHAR(100), year INT)''')
```

16.5 Read books2.csv and insert its data into the book table.

books2.csv
title,author,year
The Weirdstone of Brisingamen,Alan Garner,1960

The Spellman Files,Lisa Lutz,2007
Small Gods,Terry Pratchett,1992

```python
import csv
```

```python
books = [ 
    ["title", "author", "year"],
    ["The Weirdstone of Brisingamen", "Alan Garner", 1960],
    ["Perdido Street Station", "China Miéville", 2000],
    ["Thud!", "Terry Pratchett", 2005],
    ["The Spellman Files", "Lisa Lutz", 2007],
    ["Small Gods", "Terry Pratchett", 1992]
        ]   
```

created books2.csv

```python
with open("books2.csv", "wt") as fout:
    csvout = csv.writer(fout)
    csvout.writerows(books)
```

```python
with open("books2.csv", "rt") as file:
    reader = csv.reader(file)
    next(reader)
    
    for row in reader:
        if row:
            conn.execute("INSERT INTO books VALUES (?, ?, ?)", row)
conn.commit()
        
```

16.8 Use the sqlalchemy module to connect to the sqlite3 database books.db that you just made in exercise 16.4. As in 16.6, select and print the title column from the book table in alphabetical order.

```python
import sqlalchemy as sa
```

```python
conn = sa.create_engine("sqlite:///books.db").connect()
```

```python
from sqlalchemy import text
result = conn.execute(text("SELECT title FROM books ORDER BY title"))
for row in result:
        print(row[0])
```
